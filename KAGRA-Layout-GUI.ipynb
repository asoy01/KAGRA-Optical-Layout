# Open GUI

```python
#1000 mm round breadboard at the origin
BB1 = round_breadboard(1000*mm, center=BS.HRcenter + [0.2954, -22.468], name='BB1')

# Vacuum chamber and view ports
CIRC1 = Mechanics(shapes=[draw.Circle([0.0, 0.0], 1200*mm/2)],
                  name='CIRC1', attached_to=BB1, offset=[0.0, 0.0])
VP1 = Mechanics(shapes=[draw.Rectangle([0.0, 0.0], 200*mm, 400*mm)],
                name='VP1', attached_to=BB1, offset=[675*mm - 200*mm , 185*mm - 200*mm])
VP2 = Mechanics(shapes=[draw.Rectangle([-200*mm, -200*mm], 200*mm, 400*mm, angle=deg2rad(-30), pivot=[0,0])],
                name='VP2', attached_to=BB1,  offset=[690*mm*np.cos(deg2rad(-30)), 690*mm*np.sin(deg2rad(-30))])

#Steering mirror and its mount
M1 = opt.Mirror(HRcenter=BS.HRcenter + [0, -22.5],
                normAngleHR=0.0, diameter=2*inch, thickness=12*mm, name='M1')
MT1 = mirror_mount_2in(name='MT1', attached_to=M1)
P1 = pedestal(name='P1', attached_to=MT1, offset=MT1.points['post'])
FK1 = clamping_fork(name='FK1', attached_to=P1, offset=[0.0, 0.0],
                    fix_rotation=False)

layout = OpticalLayout(optics=OpticsList + [M1], sources=[input_beam],
                       mechanics=[BB1, CIRC1, VP1, VP2, MT1, P1, FK1],
                       rules=TraceRules(order=1, power_threshold=1e-3),
                       name='bKAGRA Layout')

layout.show()
```
